● Extract relevant columns: name, date_utc, success, details, rocket

```python
launch_data = launch_data[["name", "date_utc", "success", "details", "rocket"]]
```

```python
launch_data.shape
```

Convert date_utc to datetime and extract the year

```python
launch_data["date_utc"] = pd.to_datetime(launch_data["date_utc"])
```

```python
launch_data["year"] = launch_data["date_utc"].astype(str).str.split("-").str[0]
```

```python
launch_data[["date_utc", "year"]].head()

```

##Step 2: Load Rocket Metadata

```python
import requests
import pandas as pd

rocket_data = pd.DataFrame(requests.get("https://api.spacexdata.com/v4/rockets").json()
)
```

```python
rocket_data.columns
```

```python
rocket_data = rocket_data[["id", "name", "type", "active", "stages"]]

```

```python
rocket_data.head()
```

```python
launch_data.head()
```

#Step 3: Merge Launch and Rocket Data

```python
launches_data = launch_data.rename(columns={'name': 'launch_name'})
rockets_data = rocket_data.rename(columns={'name': 'rocket_name'})
```

```python
merged_df = pd.merge(
    launches_data,
    rockets_data,
    left_on='rocket',
    right_on='id',
    how='left'
)
(merged_df)
```

##Add Simulated Country Information


 Add a new column country and randomly assign one of these values: ['USA', 'Russia', 'India', 'China', 'France']

```python
countries = ['USA', 'Russia', 'India', 'China', 'France']
```

```python
merged_df['country'] = np.random.choice(countries, size=len(merged_df))
```

```python
(merged_df.head())
```

##Step 5: Store Merged Data in SQLite3

Use sqlite3 to create a connection and save the merged DataFrame as a table named launches

```python
conn = sqlite3.connect('launches.db')
```

```python
conn = sqlite3.connect('launches.db')
```

```python
df_from_db = pd.read_sql('SELECT * FROM launches LIMIT 5', conn)
df_from_db
```

##Step 6: Run SQL Queries on the Data to analyze

1. Launches by Country

```python
cursor=conn.cursor()
```

```python
cursor.execute("""
     SELECT country, COUNT(*) AS launches_count
    FROM launches
    GROUP BY country
    ORDER BY launches_count DESC;
    """
)
```

```python

rows=(cursor.fetchall())
for row in rows:
  print(row)
```

2. Which year had the highest number of launches?

```python
cursor.execute("""
    SELECT year, COUNT(*) AS launches_count
    FROM launches
    GROUP BY year
    ORDER BY launches_count DESC
    LIMIT 1;
""")
```

```python
row = cursor.fetchone()
```

```python
print(row)
```

3. Top 5 Missions by Launch Count

```python
cursor.execute("""
    SELECT launch_name, COUNT(*) AS launch_count
    FROM launches
    GROUP BY launch_name
    ORDER BY launch_count DESC
    LIMIT 5;
""")
```

```python
rows = cursor.fetchall()
for row in rows:
  print(row)
```
