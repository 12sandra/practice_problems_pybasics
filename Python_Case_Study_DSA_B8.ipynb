# Q1. Write a function that return the range (maximum and minimun) of a given list of numbers

```python
def display_range(list_of_numbers):
  maximum=list_of_numbers[0]                   # setting the first value of list as minimum value
  minimum=list_of_numbers[0]                   # setting the first value of list as maximum value


  for num in list_of_numbers :
    if num>= maximum:                       #if the first number is greater than maximum, reassign maximum
      maximum=num
    if num<minimum:                         ##if the first number is lesser than minimum, reassign minimum
      minimum=num
  return(minimum,maximum)

numbers=[1,5,7,2]
range= display_range(numbers)
print(range)



```

## Q2. Write function that returns mean value of a list of numbers

```python
def meancalculation(list1):
    total = 0
    length = len(list1)
    for i in list1:
        total += i
        mean = total / length
    return mean

number_list = [1, 7, 9]
print(round(meancalculation(number_list)))
```

# Q3. Write a function that returns the median value of a list of numbers

## Q4. Write a function to return the mod of a list of numbers, if no mod exists, it should return None. If more than one mod exist, return the mod values as list

```python
def ModeCalculation(numbers_list):
  frequency={}
  mode_values=[]
  highest=0

  for num in numbers_list:
    if num in frequency:
      frequency[num]+=1
    else:
      frequency[num]=1

  for count in frequency.values():
    if count>highest:
      highest=count
  for num, count in frequency.items():
      if count == highest:
            mode_values.append(num)

  return mode_values





numbers = [2, 3, 2, 4, 3, 2, 5, 3]
print(ModeCalculation(numbers))
    #my_dict=dict(numbers_list)


```
